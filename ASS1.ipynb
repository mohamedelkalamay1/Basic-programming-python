## 1. Password Validator

```python
def password_validator():
    password = input("Enter your password: ")

    has_min_length = len(password) >= 8
    has_digit = False
    has_upper = False

    for char in password:
        if char.isdigit():
            has_digit = True
        if char.isupper():
            has_upper = True

    if has_min_length and has_digit and has_upper:
        print("Valid Password")
    else:
        print("Invalid Password")


password_validator()
```

## 2. Count Characters in a String

```python
def count_characters():
    sentence = input("Enter a sentence: ")

    vowels_count = 0
    consonants_count = 0
    digits_count = 0
    spaces_count = 0

    vowels = "aeiouAEIOU"

    for char in sentence:
        if char.isalpha():
            if char in vowels:
                vowels_count += 1
            else:
                consonants_count += 1
        elif char.isdigit():
            digits_count += 1
        elif char == " ":
            spaces_count += 1

    print(f"Vowels: {vowels_count}")
    print(f"Consonants: {consonants_count}")
    print(f"Digits: {digits_count}")
    print(f"Spaces: {spaces_count}")


count_characters()
```

## 3. Student Grades List

```python
def student_grades_list():
    grades = [75, 48, 90, 62, 35, 88, 51]

    passed_count = 0
    failed_count = 0

    for grade in grades:
        print(grade)
        if grade >= 50:
            passed_count += 1
        else:
            failed_count += 1

    print(f"Passed: {passed_count}")
    print(f"Failed: {failed_count}")


student_grades_list()
```

## 4. Find the Largest and Smallest Number

```python
def largest_and_smallest():
    numbers = [15, 42, 8, 91, 23, 4, 55]

    largest = numbers[0]
    smallest = numbers[0]

    for number in numbers:
        if number > largest:
            largest = number
        if number < smallest:
            smallest = number

    print(f"Largest: {largest}")
    print(f"Smallest: {smallest}")


largest_and_smallest()
```

## 5. Word Frequency Dictionary

```python
def word_frequency():
    sentence = input("Enter a sentence: ")
    words = sentence.lower().split()

    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    print(frequency)


word_frequency()
```

## 6. Remove Duplicate Values

```python
def remove_duplicates():
    numbers = [5, 2, 5, 8, 2, 10, 8, 12]

    unique_numbers = []

    for number in numbers:
        if number not in unique_numbers:
            unique_numbers.append(number)

    print(unique_numbers)


remove_duplicates()
```

## 7. Group Words by First Letter

```python
def group_words_by_first_letter():
    words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]

    grouped = {}

    for word in words:
        first_letter = word[0]
        if first_letter in grouped:
            grouped[first_letter].append(word)
        else:
            grouped[first_letter] = [word]

    print(grouped)


group_words_by_first_letter()
```

## 8. Shopping Cart

```python
def shopping_cart():
    cart = {
        "Laptop": 25000,
        "Mouse": 500,
        "Keyboard": 1200,
        "Headphones": 900
    }

    total_price = 0
    expensive_items_count = 0

    for product, price in cart.items():
        print(f"{product}: {price}")
        total_price += price
        if price > 1000:
            expensive_items_count += 1

    print(f"Total: {total_price}")
    print(f"Products over 1000: {expensive_items_count}")

    if total_price > 25000:
        print("Expensive Order")
    else:
        print("Normal Order")


shopping_cart()
```

## 9. Common Values Between Two Lists

```python
def common_values():
    list1 = [1, 2, 3, 4, 5, 8]
    list2 = [3, 4, 5, 6, 7, 8]

    common = []

    for value in list1:
        if value in list2:
            common.append(value)

    common_tuple = tuple(common)
    print(common_tuple)


common_values()
```

## 10. Student Management

```python
def student_management():
    students = {
        "Ahmed": [70, 80, 90],
        "Sara": [95, 88, 92],
        "Omar": [40, 55, 45],
        "Mona": [60, 75, 70]
    }

    passed_students = []
    failed_students = []

    for name, grades in students.items():
        average = sum(grades) / len(grades)
        print(f"{name}: {average:.2f}")

        if average >= 50:
            print("Passed")
            passed_students.append(name)
        else:
            print("Failed")
            failed_students.append(name)

    print(f"Passed students: {passed_students}")
    print(f"Failed students: {failed_students}")


student_management()
```
